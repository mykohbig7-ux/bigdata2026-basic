```python
import pandas as pd
import matplotlib as plt
```

```python
# 데이터 셋 불러오기
df = pd.read_csv('연습문제10-4.csv')
print(df.head())
```

```python
# 'dt' 컬럼의 데이터 타입 변경
df['dt'] = pd.to_numeric(df['dt'], errors='coerce')
print(df.info())
```

```python
# 'City'와 'Country' 컬럼의 결측값 대체
print(df.isnull().sum())
df[['City']]= df[['City']].fillna('Country')
print(df.isnull().sum())
```

```python
# 'AverageTemperature' 컬럼의 결측값 대체
df['AverageTemperature'] = df['AverageTemperature'].fillna(df['AverageTemperature'].mean())
```

```python
# 'AverageTemperature' 컬럼에서 이상치 확인
plt.boxplot(df.AverageTemperature)
# 'AverageTemperature' 컬럼에서 이상치 제거
Q1 = df['AverageTemperature'].quantile(0.25)
Q3 = df['AverageTemperature'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df = df[()]
# 'AverageTemperature' 컬럼에서 이상치 다시 확인
plt.boxplot(df.AverageTemperature)
```
